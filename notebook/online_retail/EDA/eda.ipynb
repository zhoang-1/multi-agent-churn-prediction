```python
""" EDA for Online Retail Dataset """
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from churn_prediction.paths import INTERIM_NOTEBOOK_ONLINE_RETAIL_DIR, ONLINE_RETAIL_RAW_DIR
```

```python
df_09_10 = pd.read_excel(
    ONLINE_RETAIL_RAW_DIR / "online_retail_II.xlsx", sheet_name="Year 2009-2010" 
)
df_10_11 = pd.read_excel(
    ONLINE_RETAIL_RAW_DIR / "online_retail_II.xlsx", sheet_name="Year 2010-2011"
)
```

```python
print(df_09_10.columns)
print(df_10_11.columns)
print(df_09_10.shape)
print(df_10_11.shape)
```

```python
# Gộp dữ liệu
df = pd.concat(
    [df_09_10, df_10_11],
    ignore_index=True
)
```

```python
print(df.shape)
df.head(5)
```

```python
df.info()
```

```python
df.isnull().sum()
df.duplicated().sum()
```

```python
df['Invoice'] = df['Invoice'].astype(str)
df['StockCode'] = df['StockCode'].astype(str)
df['Customer ID'] = df['Customer ID'].astype(str)
df['Description'] = df['Description'].astype(str)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
# Số khách hàng
df["Customer ID"].nunique()
# Số hóa đơn
df["Invoice"].nunique()
```

```python
df.to_parquet(
    INTERIM_NOTEBOOK_ONLINE_RETAIL_DIR/"online_retail_raw.parquet",
    index=False
)
print("Saved preprocessed data to parquet file.")
```
